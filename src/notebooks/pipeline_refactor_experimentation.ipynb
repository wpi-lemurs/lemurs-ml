## Example usage with the screentime window pipeline

```python
pipeline = ScreentimeModelPipeline(
    target_type='sleep',
    time_windows=[3,6,9,12],
    save_confusion_matrices=True
)
results = pipeline.fit_predict();
```

## View Processed Screentime Data

After `fit_predict()` runs, the processed data is stored in `pipeline.processed_data_`.
This is a dictionary where keys are time windows (hours) and values are DataFrames with screentime features.

```python
# View available time windows
print("Available time windows:", list(pipeline.processed_data_.keys()))
print()

# Display data for each time window
for window, df in pipeline.processed_data_.items():
    print(f"\n{'='*80}")
    print(f"SCREENTIME DATA - {window} HOUR WINDOW")
    print(f"{'='*80}")
    print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")
    print(f"\nFirst 20 rows:")
    display(df.head(20))
```

## View Specific Time Window

You can also access a specific time window directly:

```python
# Example: View 6-hour window data
window_6h = pipeline.processed_data_[6]

print(f"6-hour window data shape: {window_6h.shape}")
print(f"\nLabel distribution:")
print(window_6h['sleep_label'].value_counts())
print(f"\nSample of hourly screentime features:")
display(window_6h[['app_user_id', 'sleep_label', 'hour_0', 'hour_1', 'hour_2', 'hour_3', 'hour_4', 'hour_5']].head(10))
```

## Screentime Subwindow Experimentation

This implementation COMBINES traditional hourly screentime features WITH
app category-based sub-window features for richer model input.

```python
subwindow_pipeline = ScreentimeModelPipeline(
    target_type='sleep',
    use_subwindows=True,        # Enable sub-window mode (ADDS to hourly features)
    lookback_hours=12,           # 12-hour lookback
    subwindow_hours=3,           # 3-hour sub-windows
    save_confusion_matrices=True
)
subwindow_results = subwindow_pipeline.fit_predict();
```

```python
# Get the combined features (dict with lookback_hours as key)
combined_data = subwindow_pipeline.processed_data_[12]

print(f"Combined features shape: {combined_data.shape}")
print(f"\nColumns ({len(combined_data.columns)} total):")

# Count feature types
hourly_cols = [c for c in combined_data.columns if c.startswith('hour_')]
subwindow_cols = [c for c in combined_data.columns if c.startswith(('most_used', 'num_apps'))]

print(f"  - Hourly features: {len(hourly_cols)}")
print(f"  - Sub-window features: {len(subwindow_cols)}")
print(f"  - Total features: {len(hourly_cols) + len(subwindow_cols)}")

print(f"\nAll columns:")
print(list(combined_data.columns))

print(f"\nFirst 10 rows:")
display(combined_data.head(10))
```

Model appears to be more accurate with the combined features
